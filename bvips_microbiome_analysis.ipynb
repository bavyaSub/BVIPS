```python
# Diversity Analysis
```

```python
from skbio.diversity import alpha_diversity
import numpy as np
import pandas as pd
```

```python
abundance_samples = relative_abundance.T
```

```python
shannon = alpha_diversity(
    metric="shannon",
    counts=abundance_samples,
    ids=abundance_samples.index
)
```

```python
shannon = pd.Series(shannon, name="Shannon")
```

```python
simpson = alpha_diversity(
    metric="simpson",
    counts=abundance_samples,
    ids=abundance_samples.index
)

simpson = pd.Series(simpson, name="Simpson")
```

```python
richness = (abundance_samples > 0).sum(axis=1)

richness.name = "Richness"
```

```python
evenness = shannon / np.log(richness)

evenness.name = "Evenness"
```

```python
dominance = abundance_samples.max(axis=1)

dominance.name = "Dominance"
```

```python
ecological_features = pd.concat(
    [
        shannon,
        simpson,
        richness,
        evenness,
        dominance
    ],
    axis=1
)
```

```python
ecological_features.head()
```

```python
from scipy.spatial.distance import braycurtis
```

```python
dysbiosis_bray = {}

for sample in relative_abundance.columns:
    dysbiosis_bray[sample] = braycurtis(
        relative_abundance[sample],
        healthy_reference
    )

dysbiosis_bray = pd.Series(
    dysbiosis_bray,
    name="Bray_Curtis_Dysbiosis"
)
```

```python
# Inflammatory Feature Engineering
```

```python
inflammatory_annotations = pd.DataFrame(
    index=relative_abundance.index
)

inflammatory_annotations["LPS"] = 0
inflammatory_annotations["Pathogen"] = 0
inflammatory_annotations["Protease"] = 0
```

```python
microbiome = relative_abundance.T

microbiome.head()
```

```python
inflammatory_features = pd.DataFrame(
    index=microbiome.index
)

inflammatory_features.head()
```

```python
healthy_center = microbiome.loc[healthy_samples].mean(axis=0)

healthy_center.head()
```

```python
from scipy.spatial.distance import euclidean

dysbiosis_scores = []

for patient in microbiome.index:
    distance = euclidean(
        microbiome.loc[patient],
        healthy_center
    )
    dysbiosis_scores.append(distance)

inflammatory_features["Dysbiosis"] = dysbiosis_scores
```

```python
protease_scores = annotations.set_index("Species")["Protease"]
```

```python
protease_weighted = microbiome.mul(protease_scores, axis=1)
```

```python
# B-VIPS Score Calculation
```

```python
scaled_features = pd.DataFrame(
    scaled_features,
    columns=[
        "Dysbiosis_component", 
        "LPS_component", 
        "Pathogen_component", 
        "Protease_component"
    ],index=features.index
)

scaled_features.head()
```

```python
scaled_features["BVIPS"] = (
    scaled_features["Dysbiosis_component"] +
    scaled_features["LPS_component"] +
    scaled_features["Pathogen_component"] +
    scaled_features["Protease_component"]
) / 4
```

```python
results = scaled_features.copy()

results["Cluster"] = inflammatory_features["Cluster"]

results.head()
```

```python
results = results.rename(columns={"Dysbiosis_component": 'Dysbiosis_score', 'LPS_component': 'LPS_score', "Pathogen_component": 'Pathogen_score', "Protease_component": 'Protease_score'})
```
